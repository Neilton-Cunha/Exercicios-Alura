 pedir para uma IA gerar, em Python, uma lista de 20 e-mails

 criar uma função que receba essa lista de 20 e-mails como argumento. A função deve percorrer a lista e, para cada e-mail, chamar a IA do Gemini via API para resumir o conteúdo do e-mail em uma linha, definindo o que a pessoa queria no e-mail.

```python
emails = [
    "Olá, gostaria de saber mais informações sobre os planos de internet disponíveis para minha região.",
    "Bom dia, estou entrando em contato para solicitar a segunda via da minha fatura.",
    "Prezados, meu pedido ainda não foi entregue e gostaria de acompanhar o status da entrega.",
    "Gostaria de agendar uma consulta para a próxima semana, se houver horários disponíveis.",

]
```

```python
def exercicio_alura(lista_de_emails):

  lista_emails = []
  for num, i in enumerate(emails):
      resposta = client.models.generate_content(
          model = 'gemini-3-flash-preview',
          contents= f'Faz o resumo dos emails que serão passados pra voce em poucas palavras{i}')
      print(f'Email{num+1}: {resposta.text}')
      lista_emails.append(f'Email{num+1}: {resposta.text}')
      print('-'* 50)

  return lista_emails

lista_emails = exercicio_alura(emails)
```

```python
from groq import Groq

client_groq = Groq()
completion = client_groq.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
      {
        "role": "user",
        "content": "Em qual mes estamos hoje?"
      }
    ],
    temperature=1,
    max_completion_tokens=7000,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```

## Escrevendo em arquivo python .txt ou .csv

with open("Nome do arquivo.txt", "w" ou "a", encoding = 'utf-8') as nome que quiser:

Use o modo **w** quando quiser reescrever o arquivo do zero, descartando tudo que havia antes.

Use o modo **a** quando quiser manter o conteúdo anterior e apenas adicionar novas linhas ao final.

```python
with open("Lista de email.txt", "w" , encoding = 'utf-8') as emails:
    for resumo in lista_emails:
       emails.write (resumo + '\n')
       # esse write pode ser ele proprio onde vai ser lido cada documento e impresso individualmente
       #sendo necessario a implementação do for para ler mais de um doc
       # enquanto writelines ele ler todos os documentos imprimindo em uma unica linha
```

## Lendo em arquivo python .txt ou .csv

o que diferencia é substituir o w/a por r (read)

```python
nova_lista_de_leitura = []

with open("Lista de email.txt", "r" , encoding = 'utf-8') as emails:
    for resumo in emails:
       nova_lista_de_leitura.append(resumo.strip())
```

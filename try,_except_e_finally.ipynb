Uso de blocos try, except e finally

```python
try:
  #try para tentar executar um bloco de código.
    pessoas = 6
    pessoas_2 = "Mário, João, Silvia, Maria"
    pessoas + pessoas_2
except ValueError as ve:
  #except captura o tipo específico de erro ValueError, TypeError. etc
    print("Você não digitou um número inteiro válido")
    print("Detalhes do erro:", ve)
except TypeError as te:
    print("Você não pode somar um número inteiro com uma string")
    print("Detalhes do erro:", te)
except Exception as e:
    print("Detalhes do erro:", e)
finally:
    print("Finalizei o processo")
```

```python
perguntas = [
    "O que é uma escala musical?",
    "Qual a diferença entre tom e semitom?",
    "O que forma um acorde maior?",
    "O que é um compasso na música?",
    "Qual a função da clave de sol?",
    "O que significa andamento em uma composição?",
    "O que é uma progressão harmônica?",
    "Qual a diferença entre melodia e harmonia?",
    "O que é um intervalo musical?",
    "Quais são as notas da escala de Dó Maior?"
]
```

```python
with open("Perguntas de musica.txt", "w" , encoding = 'utf-8') as desafio:
   for exercicio in perguntas:
      desafio.write (exercicio +'\n')
```

```python
nova_lista_de_perguntas = []

with open("Perguntas de musica.txt", "r" , encoding = 'utf-8') as desafio:
  for exercicio in desafio:
      nova_lista_de_perguntas.append(exercicio.strip())
```

```python
from groq import Groq

respostas = []

client_groq = Groq()
completion = client_groq.chat.completions.create(
      model="openai/gpt-oss-20b",
      messages=[
      {
        "role": "user",
        "content":f'responde de forma sucinta essas perguntas, mas so reotrne as respostas, sem enumera-las: {perguntas}'
      }
    ],
    temperature=1,
    max_completion_tokens=7000,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

resposta = ''

for chunk in completion:
    resposta += chunk.choices[0].delta.content or ""

respostas.append(resposta)
```

```python
respostas_separadas = []

for linha in respostas[0].split('\n'):
    if linha.strip():
        respostas_separadas.append(linha.strip())
```

```python
dicionario = {'Perguntas': nova_lista_de_perguntas, 'Respostas': respostas_separadas}
df = pd.DataFrame(dicionario)
df.to_csv("Perguntas de musica.csv", index=False, encoding="utf-8")
```
